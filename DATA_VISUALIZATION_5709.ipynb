```python
import pandas as pd

# Specify the path to your CSV file and the encoding
path = '/content/Global_Education.csv'
encoding = 'ISO-8859-1'

# Read the CSV file into a pandas DataFrame
edu = pd.read_csv(path, encoding=encoding)
print('Data Dimensions are:', edu.shape)
```

```python
india = edu.loc[edu['Countries and areas'] == 'India']
japan = edu.loc[edu['Countries and areas'] == 'Japan']
southkorea = edu.loc[edu['Countries and areas'] == 'South Korea']
russia = edu.loc[edu['Countries and areas'] == 'Russia']
switzerland = edu.loc[edu['Countries and areas'] == 'Switzerland']
australia = edu.loc[edu['Countries and areas'] == 'Australia']
```

```python
countries = ['India', 'Japan', 'South Korea', 'Russia', 'Switzerland', 'Australia']
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x = range(len(countries))
width = 0.3
def bar_plot_male_female(male_count, female_count, title, xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(12, 6))
    fig.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)
    plt.bar(x, male_count, width, label='Male', color='Blue')
    plt.bar([i + width for i in x], female_count, width, label='Female', color='Pink')
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.xticks([i + width / 2 for i in x], countries)
    plt.legend()
    plt.show()
male_count_OOSR_Pre0Primary_Age_Male = [india['OOSR_Pre0Primary_Age_Male'].values[0],
               japan['OOSR_Pre0Primary_Age_Male'].values[0],
               southkorea['OOSR_Pre0Primary_Age_Male'].values[0],
               russia['OOSR_Pre0Primary_Age_Male'].values[0],
               switzerland['OOSR_Pre0Primary_Age_Male'].values[0],
               australia['OOSR_Pre0Primary_Age_Male'].values[0]]

female_count_OOSR_Pre0Primary_Age_Male = [india['OOSR_Pre0Primary_Age_Female'].values[0],
               japan['OOSR_Pre0Primary_Age_Female'].values[0],
               southkorea['OOSR_Pre0Primary_Age_Female'].values[0],
               russia['OOSR_Pre0Primary_Age_Female'].values[0],
               switzerland['OOSR_Pre0Primary_Age_Female'].values[0],
               australia['OOSR_Pre0Primary_Age_Female'].values[0]]

bar_plot_male_female(male_count_OOSR_Pre0Primary_Age_Male, female_count_OOSR_Pre0Primary_Age_Male, 'Out-of-school rate for pre-primary age', 'Country', 'OOSR_Pre0Primary_Age_Male')
```

```python
def bar_plot_read_math(read, math, title, xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(12, 6))
    fig.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)
    plt.bar(x, read, width, label='Reading_proficiency', color='red')
    plt.bar([i + width for i in x], math, width, label='Math_proficiency', color='blue')
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.xticks([i + width / 2 for i in x], countries)
    plt.legend()
    plt.show()
reading_proficiency2_3 = [india['Grade_2_3_Proficiency_Reading'].values[0],
               japan['Grade_2_3_Proficiency_Reading'].values[0],
               southkorea['Grade_2_3_Proficiency_Reading'].values[0],
               russia['Grade_2_3_Proficiency_Reading'].values[0],
               switzerland['Grade_2_3_Proficiency_Reading'].values[0],
               australia['Grade_2_3_Proficiency_Reading'].values[0]]

math_proficiency2_3 = [india['Grade_2_3_Proficiency_Math'].values[0],
               japan['Grade_2_3_Proficiency_Math'].values[0],
               southkorea['Grade_2_3_Proficiency_Math'].values[0],
               russia['Grade_2_3_Proficiency_Math'].values[0],
               switzerland['Grade_2_3_Proficiency_Math'].values[0],
               australia['Grade_2_3_Proficiency_Reading'].values[0]]

bar_plot_read_math(reading_proficiency2_3, math_proficiency2_3, 'Proficiency in reading and math for grade 2-3 students', 'Country', 'Grade_2_3_Proficiency')
```

```python
def birth(data_values, title, xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(8, 6))
    plt.bar(x, data_values, color='green')
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.xticks(x, countries)
    plt.show()
birth_rate = [india['Birth_Rate'].values[0],
               japan['Birth_Rate'].values[0],
               southkorea['Birth_Rate'].values[0],
               russia['Birth_Rate'].values[0],
               switzerland['Birth_Rate'].values[0],
               australia['Birth_Rate'].values[0]]
birth(birth_rate, 'Birth rate in the respective countries', 'Country', 'Birth_Rate')
```

```python
def unemployment(data_values, title, xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(8, 6))
    plt.bar(x, data_values, color='purple')
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.xticks(x, countries)
    plt.show()
unemployment_rate = [india['Unemployment_Rate'].values[0],
               japan['Unemployment_Rate'].values[0],
               southkorea['Unemployment_Rate'].values[0],
               russia['Unemployment_Rate'].values[0],
               switzerland['Unemployment_Rate'].values[0],
               australia['Unemployment_Rate'].values[0]]
unemployment(unemployment_rate, 'Unemployment rate in the respective countries', 'Country', 'Unemployment_Rate')
```

```python
import pandas as pd

gross_primary_education_enroll = [india['Gross_Primary_Education_Enrollment'].values[0],
               japan['Gross_Primary_Education_Enrollment'].values[0],
               southkorea['Gross_Primary_Education_Enrollment'].values[0],
               russia['Gross_Primary_Education_Enrollment'].values[0],
               switzerland['Gross_Primary_Education_Enrollment'].values[0],
               australia['Gross_Primary_Education_Enrollment'].values[0]]

gross_tertiary_education_enroll = [india['Gross_Tertiary_Education_Enrollment'].values[0],
               japan['Gross_Tertiary_Education_Enrollment'].values[0],
               southkorea['Gross_Tertiary_Education_Enrollment'].values[0],
               russia['Gross_Tertiary_Education_Enrollment'].values[0],
               switzerland['Gross_Tertiary_Education_Enrollment'].values[0],
               australia['Gross_Tertiary_Education_Enrollment'].values[0]]
fig, ax = plt.subplots(figsize=(12, 6))
fig.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)
x = range(len(countries))
width = 0.4
plt.bar(x, gross_primary_education_enroll, width, label='Primary', color='orange')
plt.bar([i + width for i in x], gross_tertiary_education_enroll, width, label='Tertiary', color='blue')
plt.xlabel('Country')
plt.ylabel('Gross enrollment')
plt.title('Gross enrollment in primary and tertiary education')
plt.xticks([i + width / 2 for i in x], countries)
plt.legend()
plt.show()
```
